```python
import os
from google.colab import files
os.makedirs('inputs', exist_ok=True)
up = files.upload()
for name, data in up.items():
    open(os.path.join('inputs', name), 'wb').write(data)
print(sorted(os.listdir('inputs')))
```

```python
# T4 has 15 GB: the default bf16 pipeline needs 12-16 GB at 1280, so run with --group_offload at 1024.
!python inference/scripts/inference_psd.py --srcp inputs/ --save_dir outputs --save_to_psd --resolution 1024 --group_offload
!find outputs -maxdepth 2 | head -40
```
